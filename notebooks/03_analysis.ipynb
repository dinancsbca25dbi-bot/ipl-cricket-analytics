# Section L

```python
#L1
q("""
SELECT m.season_year,                        -- what to show
ROUND(AVG(i.runs),1) AS               -- round it off
avg_innings_score
FROM   matches_clean m                       -- one row = one match
JOIN   v_innings i ON i.match_id=m.match_id  -- attach the other table
GROUP  BY m.season_year
ORDER  BY m.season_year DESC
LIMIT  5;
""") 
```

```python
#L2
q("""      
SELECT m.season_year,                     -- what to show
       ROUND(100.0*SUM(CASE WHEN          -- round it off
              b.batter_runs=6 THEN 1 ELSE 0
              END)/SUM(b.is_legal),2) AS  -- branches of the CASE
              six_pct
FROM   v_ball b                           -- one row = one ball
JOIN   matches_clean m ON                 -- attach the other table
       m.match_id=b.match_id
GROUP  BY m.season_year                   -- one row per group
ORDER  BY m.season_year DESC              -- sort
LIMIT  5;
""")
```

```python
#L3
q("""
SELECT m.season_year,                     -- what to show
       ROUND(100.0*SUM(CASE WHEN          -- round it off
              b.total_runs=0 AND
              b.is_legal=1 THEN 1 ELSE 0
              END)/SUM(b.is_legal),2) AS  -- branches of the CASE
              dot_pct
FROM   v_ball b                           -- one row = one ball
JOIN   matches_clean m ON                 -- attach the other table
       m.match_id=b.match_id
GROUP  BY m.season_year                   -- one row per group
ORDER  BY m.season_year DESC              -- sort
LIMIT  5;  
""")
```

```python
#L4
q("""
SELECT season_year, COUNT(*) AS matches  -- row count
FROM   matches_clean                     -- one row = one match
GROUP  BY season_year
ORDER  BY matches DESC
LIMIT  5;
""")
```

```python
#L5
q("""
SELECT m.season_year,                                  -- what to show
ROUND(6.0*SUM(b.total_runs)/SUM(b.is_legal),2)  -- round it off
AS death_run_rate
FROM   v_ball b
JOIN   matches_clean m ON
m.match_id=b.match_id
WHERE  b.phase='Death'
GROUP  BY m.season_year
LIMIT  5;  
""")
```

```python
#L6
q("""
WITH per_match AS (SELECT match_id,  -- name a first query
       SUM(total_runs) AS runs FROM v_ball
       GROUP BY match_id)            -- one row per group
SELECT m.season_year,                -- what to show
       ROUND(AVG(p.runs),1) AS       -- round it off
              avg_match_runs
FROM   per_match p                   -- the source table
JOIN   matches_clean m ON            -- attach the other table
       m.match_id=p.match_id
GROUP  BY m.season_year              -- one row per group
ORDER  BY avg_match_runs DESC        -- sort
LIMIT  5;  
""")
```

```python
#L7
q("""
SELECT m.season_year,                     -- what to show
       ROUND(100.0*SUM(CASE WHEN          -- round it off
              b.bowler_type LIKE
              '%Legbreak%' OR b.bowler_type
              LIKE '%Offbreak%' OR
              b.bowler_type LIKE
              '%Orthodox%' OR b.bowler_type
              LIKE '%Googly%' OR b.bowler_type
              LIKE '%Wrist spin%' THEN
              b.is_legal ELSE 0
              END)/SUM(b.is_legal),1) AS  -- branches of the CASE
              pct_balls_spin
FROM   v_ball b                           -- one row = one ball
JOIN   matches_clean m ON                 -- attach the other table
       m.match_id=b.match_id
GROUP  BY m.season_year                   -- one row per group
ORDER  BY m.season_year DESC              -- sort
LIMIT  5; 
""")
```

```python
#L8
q("""
WITH per_season AS ( SELECT m.season_year AS  -- name a first query
       yr, AVG(i.runs) AS avg_runs FROM
       matches_clean m JOIN v_innings i ON
       i.match_id=m.match_id GROUP BY
       m.season_year)
SELECT yr,                                    -- what to show
       ROUND(avg_runs,1) AS avg_runs,         -- round it off
       ROUND(avg_runs - LAG(avg_runs) OVER    -- previous row's value
              (ORDER BY yr),1) AS
              change_from_last_season
FROM   per_season                             -- the source table
ORDER  BY yr DESC                             -- sort
LIMIT  5; 
""")
```

# Section M

```python
#M1
q("""
SELECT batter,                                        -- what to show
       SUM(is_legal) AS balls,                        -- sample size
       ROUND(100.0*SUM(batter_runs)/SUM(is_legal),2)  -- runs per 100 bal
              AS sr_pp,
       SUM(CASE WHEN player_out = batter
              THEN 1 ELSE 0 END) AS                   -- branches of the
              times_out_in_pp
FROM   v_ball                                         -- one row = one ba
WHERE  phase='Powerplay'                              -- powerplay only
GROUP  BY batter                                      -- one row per group
HAVING balls >= 600 AND sr_pp >= 140                  -- filter the groups
ORDER  BY sr_pp DESC                                  -- sort
LIMIT  5; 
""")
```

```python
#M2
q("""
SELECT b.batter AS player,             -- what to show
       SUM(b.batter_runs) AS runs,
       (SELECT SUM(x.bowler_wicket) FROM
              v_ball x WHERE x.bowler =
              b.batter) AS wickets
FROM   v_ball b                        -- one row = one ball
GROUP  BY b.batter                     -- one row per group
HAVING runs >= 1000 AND wickets >= 50  -- filter the groups
ORDER  BY runs DESC                    -- sort
LIMIT  8;   
""")
```

```python
#M3
q("""
SELECT bowler,                                -- what to show
       ROUND(6.0*SUM(CASE WHEN                -- round it off
              phase='Powerplay' THEN
              bowler_runs ELSE 0 END)
              /NULLIF(SUM(CASE WHEN           -- divide, no zero risk
              phase='Powerplay' THEN
              is_legal ELSE 0 END),0),2) AS
              pp,
       ROUND(6.0*SUM(CASE WHEN                -- round it off
              phase='Middle' THEN
              bowler_runs ELSE 0 END)
              /NULLIF(SUM(CASE WHEN           -- divide, no zero risk
              phase='Middle' THEN is_legal
              ELSE 0 END),0),2) AS mid,       -- branches of the CASE
       ROUND(6.0*SUM(CASE WHEN phase='Death'  -- round it off
              THEN bowler_runs ELSE 0 END)    -- branches of the CASE
              /NULLIF(SUM(CASE WHEN           -- divide, no zero risk
              phase='Death' THEN is_legal
              ELSE 0 END),0),2) AS death      -- branches of the CASE
FROM   v_ball                                 -- one row = one ball
GROUP  BY bowler                              -- one row per group
HAVING SUM(CASE WHEN phase='Powerplay' THEN   -- filter the groups
       is_legal ELSE 0 END) >= 300 AND
       SUM(CASE WHEN phase='Middle' THEN      -- if-then-else per row
       is_legal ELSE 0 END) >= 300 AND
       SUM(CASE WHEN phase='Death' THEN       -- if-then-else per row
       is_legal ELSE 0 END) >= 300 AND pp <
       8 AND mid < 8 AND death < 9
ORDER  BY (pp + mid + death) ASC              -- sort
LIMIT  5;
""")
```

```python
#M4
q("""
SELECT batter,                                        -- what to show
       SUM(is_legal) AS balls,                        -- sample size
       ROUND(100.0*SUM(batter_runs)/SUM(is_legal),2)  -- runs per 100 bal
              AS sr
FROM   v_ball                                         -- one row = one ba
WHERE  innings=2 AND phase='Death'                    -- death overs only
GROUP  BY batter                                      -- one row per group
HAVING balls >= 200                                   -- filter the groups
ORDER  BY sr DESC                                     -- sort
LIMIT  5;   
""")
```

```python
#M5
q("""    
SELECT bowler,                                      -- what to show
SUM(is_legal) AS balls,                      -- sample size
ROUND(6.0*SUM(bowler_runs)/SUM(is_legal),2)  -- runs per over conc
AS economy
FROM   v_ball
WHERE  phase='Powerplay' AND (bowler_type
LIKE '%Legbreak%' OR bowler_type LIKE
'%Offbreak%' OR bowler_type LIKE
'%Orthodox%' OR bowler_type LIKE
'%Googly%' OR bowler_type LIKE
'%Wrist spin%')
GROUP  BY bowler
HAVING balls >= 200                                 -- one row = one ball -- powerplay only
  -- one row per group -- filter the groups
ORDER  BY economy ASC                               -- sort
LIMIT  5; """)
```

```python
#M6
q("""
SELECT b.batter,
    SUM(b.is_legal) AS balls,
    ROUND(100.0 * SUM(b.batter_runs) / SUM(b.is_legal), 2) AS sr_here
FROM v_ball b
JOIN matches_clean m
  ON m.match_id = b.match_id
WHERE m.venue = 'Wankhede Stadium'
GROUP BY b.batter
HAVING balls >= 200
ORDER BY sr_here DESC
LIMIT 5;
""")
```

```python
#M7
q("""
SELECT b.batter,                                          -- what to show
       SUM(b.batter_runs) AS middle_runs,
       ROUND(100.0*SUM(b.batter_runs)/SUM(b.is_legal),2)  -- round it off
              AS sr,
       SUM(CASE WHEN b.player_out = b.batter
              THEN 1 ELSE 0 END) AS                       -- branches of
              outs_in_middle,
       ROUND(1.0*SUM(b.batter_runs)/                      -- round it off
              NULLIF(SUM(CASE WHEN
              b.player_out = b.batter THEN 1
              ELSE 0 END),0),2) AS average                -- branches of
FROM   v_ball b                                           -- one row = on
WHERE  b.phase = 'Middle'                                 -- middle overs
GROUP  BY b.batter                                        -- one row per
HAVING SUM(b.is_legal) >= 500 AND sr >= 140               -- filter the g
       AND average >= 25                                  -- and this too
ORDER  BY sr DESC                                         -- sort
LIMIT  8; """)
```

```python
#M8
q("""
SELECT batter,                                -- what to show
       ROUND(100.0*SUM(CASE WHEN innings=1    -- round it off
              THEN batter_runs ELSE 0 END)    -- branches of the CASE
              /NULLIF(SUM(CASE WHEN           -- divide, no zero risk
              innings=1 THEN is_legal ELSE 0
              END),0),1) AS sr_first,         -- branches of the CASE
       ROUND(100.0*SUM(CASE WHEN innings=2    -- round it off
              THEN batter_runs ELSE 0 END)    -- branches of the CASE
              /NULLIF(SUM(CASE WHEN           -- divide, no zero risk
              innings=2 THEN is_legal ELSE 0
              END),0),1) AS sr_second         -- branches of the CASE
FROM   v_ball                                 -- one row = one ball
GROUP  BY batter                              -- one row per group
HAVING SUM(CASE WHEN innings=1 THEN is_legal  -- filter the groups
       ELSE 0 END) >= 800 AND SUM(CASE WHEN   -- branches of the CASE
       innings=2 THEN is_legal ELSE 0 END)
       >= 800 AND sr_first >= 135 AND
       sr_second >= 135
ORDER  BY (sr_first + sr_second) DESC         -- sort
LIMIT  5;  """)
```

```python
#M9
q("""
WITH totals AS ( SELECT batting_team,         -- name a first query
       batter, SUM(batter_runs) AS runs FROM  -- total runs
       v_ball GROUP BY batting_team, batter)
SELECT batting_team,                          -- what to show
       batter,
       runs,
       RANK() OVER (PARTITION BY              -- rank within group
              batting_team ORDER BY runs
              DESC) AS rank_in_team
FROM   totals                                 -- the source table
WHERE  batting_team = 'Chennai Super Kings'   -- filter the rows
ORDER  BY rank_in_team                        -- sort
LIMIT  5;   """)
```
